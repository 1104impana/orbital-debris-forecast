```python
recent_df = ml_df[ml_df["Year"] >= 2000].copy()

recent_df[["Year", "Debris", "Total"]]
```

```python
recent_df["Debris_Growth"] = recent_df["Debris"].diff()

recent_df[["Year", "Debris", "Debris_Growth"]].tail(15)
```

```python
recent_df["Debris_Growth_Rate"] = (
    recent_df["Debris"].pct_change() * 100
)

recent_df[["Year", "Debris", "Debris_Growth_Rate"]].tail(15)
```
